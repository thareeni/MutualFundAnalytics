```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("../data/db/bluestock_mf.db")

nav_df = pd.read_sql("SELECT * FROM fact_nav", conn)

conn.close()

print(nav_df.columns)
nav_df.head()
```

```python
nav_df["date"] = pd.to_datetime(nav_df["date"])

nav_df = nav_df.sort_values(["amfi_code", "date"])

nav_df["daily_return"] = (
    nav_df.groupby("amfi_code")["nav"]
          .pct_change()
)

nav_df.head()
```

```python
cagr_data = []

for code, group in nav_df.groupby("amfi_code"):

    group = group.sort_values("date")

    end_nav = group["nav"].iloc[-1]
    end_date = group["date"].iloc[-1]

    row = {"amfi_code": code}

    for years in [1, 3, 5]:

        start_date = end_date - pd.DateOffset(years=years)

        temp = group[group["date"] >= start_date]

        if len(temp) > 1:
            start_nav = temp["nav"].iloc[0]
            cagr = (end_nav / start_nav) ** (1 / years) - 1
            row[f"CAGR_{years}Y"] = cagr * 100
        else:
            row[f"CAGR_{years}Y"] = np.nan

    cagr_data.append(row)

cagr_df = pd.DataFrame(cagr_data)

cagr_df.head()
```

```python
cagr_df.to_csv("../reports/cagr.csv", index=False)
```

```python
risk_free = 0.065

sharpe = []

for code, group in nav_df.groupby("amfi_code"):

    r = group["daily_return"].dropna()

    if len(r) > 1:

        sr = ((r.mean() * 252) - risk_free) / (r.std() * np.sqrt(252))

    else:

        sr = np.nan

    sharpe.append([code, sr])

sharpe_df = pd.DataFrame(sharpe, columns=["amfi_code", "Sharpe"])

sharpe_df.sort_values("Sharpe", ascending=False).head()
```

```python
sharpe_df.to_csv("../reports/sharpe_ratio.csv", index=False)
```

```python
sortino = []

for code, group in nav_df.groupby("amfi_code"):

    r = group["daily_return"].dropna()

    downside = r[r < 0]

    if len(downside) > 1:

        ratio = ((r.mean() * 252) - risk_free) / (downside.std() * np.sqrt(252))

    else:

        ratio = np.nan

    sortino.append([code, ratio])

sortino_df = pd.DataFrame(sortino, columns=["amfi_code", "Sortino"])

sortino_df.head()
```

```python
sortino_df.to_csv("../reports/sortino_ratio.csv", index=False)
```

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("../data/db/bluestock_mf.db")

nav_df = pd.read_sql("SELECT * FROM fact_nav", conn)

conn.close()

nav_df.head()
```

```python
print(nav_df.columns.tolist())
```

```python
nav_df["date"] = pd.to_datetime(nav_df["date"])
```

```python
nav_df["date"] = pd.to_datetime(nav_df["date"])
```

```python
nav_df = nav_df.sort_values(["amfi_code", "date"])
```

```python
import numpy as np

cagr_list = []

for fund, group in nav_df.groupby("amfi_code"):

    start_nav = group.iloc[0]["nav"]
    end_nav = group.iloc[-1]["nav"]

    years = (
        group.iloc[-1]["date"] -
        group.iloc[0]["date"]
    ).days / 365.25

    if years > 0:
        cagr = ((end_nav / start_nav) ** (1 / years) - 1) * 100
    else:
        cagr = np.nan

    cagr_list.append([fund, cagr])

cagr_df = pd.DataFrame(
    cagr_list,
    columns=["amfi_code", "CAGR (%)"]
)

cagr_df.head()
```

```python
cagr_df.to_csv("../reports/cagr.csv", index=False)
```

```python
risk_free_rate = 0.065  # 6.5%

sharpe_df = (
    nav_df.groupby("amfi_code")["daily_return"]
    .agg(["mean", "std"])
    .reset_index()
)

sharpe_df["Sharpe Ratio"] = (
    (sharpe_df["mean"] * 252 - risk_free_rate)
    / (sharpe_df["std"] * (252 ** 0.5))
)

sharpe_df.head()
```

```python
nav_df.columns
```

```python
nav_df["date"] = pd.to_datetime(nav_df["date"])

nav_df = nav_df.sort_values(["amfi_code", "date"])

nav_df["daily_return"] = (
    nav_df.groupby("amfi_code")["nav"]
          .pct_change()
)
```

```python
nav_df.head()
```

```python
risk_free_rate = 0.065

sharpe_df = (
    nav_df.groupby("amfi_code")["daily_return"]
          .agg(["mean", "std"])
          .reset_index()
)

sharpe_df["Sharpe Ratio"] = (
    (sharpe_df["mean"] * 252 - risk_free_rate)
    / (sharpe_df["std"] * (252 ** 0.5))
)

sharpe_df.head()
```

```python
risk_free_rate = 0.065

sortino_list = []

for code, group in nav_df.groupby("amfi_code"):

    returns = group["daily_return"].dropna()

    downside = returns[returns < 0]

    if len(downside) > 0:
        downside_std = downside.std() * (252 ** 0.5)
    else:
        downside_std = float("nan")

    annual_return = returns.mean() * 252

    sortino = (annual_return - risk_free_rate) / downside_std

    sortino_list.append([code, sortino])

sortino_df = pd.DataFrame(
    sortino_list,
    columns=["amfi_code", "Sortino Ratio"]
)

sortino_df.head()
```

```python
mdd_list = []

for code, group in nav_df.groupby("amfi_code"):

    group = group.sort_values("date")

    group["running_max"] = group["nav"].cummax()

    group["drawdown"] = (group["nav"] / group["running_max"]) - 1

    max_dd = group["drawdown"].min()

    mdd_list.append([code, max_dd])

mdd_df = pd.DataFrame(
    mdd_list,
    columns=["amfi_code", "Maximum Drawdown"]
)

mdd_df.head()
```

```python
mdd_df.to_csv("../reports/maximum_drawdown.csv", index=False)
```

```python
scorecard = cagr_df.merge(
    sharpe_df[["amfi_code", "Sharpe Ratio"]],
    on="amfi_code"
)

scorecard = scorecard.merge(
    mdd_df,
    on="amfi_code"
)

scorecard["CAGR Rank"] = scorecard["CAGR (%)"].rank(ascending=False)
scorecard["Sharpe Rank"] = scorecard["Sharpe Ratio"].rank(ascending=False)
scorecard["MDD Rank"] = scorecard["Maximum Drawdown"].rank(ascending=False)

scorecard["Score"] = (
    0.30 * (scorecard["CAGR Rank"].max() - scorecard["CAGR Rank"]) +
    0.25 * (scorecard["Sharpe Rank"].max() - scorecard["Sharpe Rank"]) +
    0.20 * (scorecard["MDD Rank"].max() - scorecard["MDD Rank"])
)

scorecard = scorecard.sort_values("Score", ascending=False)

scorecard.head()
```

```python
scorecard.to_csv("../reports/fund_scorecard.csv", index=False)
```

```python
benchmark_df["date"] = pd.to_datetime(benchmark_df["date"])

benchmark_df = benchmark_df.sort_values(["index_name", "date"])

benchmark_df["benchmark_return"] = (
    benchmark_df.groupby("index_name")["close_value"]
    .pct_change()
)

benchmark_df.head()
```

```python
nifty50 = benchmark_df[
    benchmark_df["index_name"] == "NIFTY50"
]

nifty50.head()
```

```python
merged_df = pd.merge(
    nav_df,
    nifty50[["date", "benchmark_return"]],
    on="date",
    how="inner"
)

merged_df.head()
```

```python
tracking_error = (
    merged_df.groupby("amfi_code")
    .apply(
        lambda x: (
            (x["daily_return"] - x["benchmark_return"]).std()
            * (252 ** 0.5)
        )
    )
    .reset_index(name="Tracking Error")
)

tracking_error.head()
```

```python
tracking_error.to_csv(
    "../reports/tracking_error.csv",
    index=False
)
```

```python
from scipy.stats import linregress
```

```python
alpha_beta = []

for code, group in merged_df.groupby("amfi_code"):

    temp = group.dropna(subset=["daily_return", "benchmark_return"])

    if len(temp) > 30:

        slope, intercept, r, p, stderr = linregress(
            temp["benchmark_return"],
            temp["daily_return"]
        )

        alpha = intercept * 252
        beta = slope

        alpha_beta.append([code, alpha, beta])

alpha_beta_df = pd.DataFrame(
    alpha_beta,
    columns=["amfi_code", "Alpha", "Beta"]
)

alpha_beta_df.head()
```

```python
alpha_beta_df.to_csv(
    "../reports/alpha_beta.csv",
    index=False
)
```

```python
scorecard = cagr_df.merge(
    sharpe_df[["amfi_code", "Sharpe Ratio"]],
    on="amfi_code"
)

scorecard = scorecard.merge(
    alpha_beta_df,
    on="amfi_code"
)

scorecard.head()
```
